```python
import numpy as np

g = lambda x: -np.sin(2*x) + (5/4)*x - 3/4

gp = lambda x: -2*np.cos(2*x) + 5/4

def fixedpt(x0, tol, Nmax):
    x = x0

    for count in range(Nmax):
        x1 = g(x)

        if x1 != 0:
            err = abs(x1 - x) / abs(x1)
        else:
            err = abs(x1 - x)

        if err < tol:
            return x1, count + 1, 0

        x = x1

    return x, Nmax, 1


tol = 1e-11
Nmax = 1000

guesses = np.arange(-1.0, 7.1, 0.1)

for x0 in guesses:
    root, count, ier = fixedpt(x0, tol, Nmax)

    if ier == 0:
        print("initial guess =", x0)
        print("root =", root)
        print("iterations =", count)
        print("|g'(root)| =", abs(gp(root)))
        print()
    else:
        print(f"initial guess = {x0} did not converge")
        print()
```
