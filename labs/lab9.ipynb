```python
def eval_legendre(n, x):
    p = np.zeros(n+1)
    p[0] = 1

    if n == 0:
        return p

    p[1] = x

    for i in range(1, n):
        p[i+1] = ((2*i+1)*x*p[i] - i*p[i-1]) / (i+1)

    return p
```
